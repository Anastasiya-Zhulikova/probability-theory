Графически показать стремление частотной вероятности к классической при увеличении числа опытов.

<b>Решение</b>. Приведу пример задачи:
> У игрока на столе колода из 36 карт. Он выбирает из колоды ровно две карты. С какой вероятностью он вытащит два чёрных туза с первого раза?

```python
from itertools import permutations

#Первая буква - цвет (red or black), вторая - карта
cards = ["r6", "r6", "b6", "b6",
        "r7", "r7", "b7", "b7",
        "r8", "r8", "b8", "b8",
        "r9", "r9", "b9", "b9",
        "r10", "r10", "b10", "b10",
        "rV", "rV", "bV", "bV",
        "rD", "rD", "bD", "bD",
        "rK", "rK", "bK", "bK",
        "rT", "rT", "bT", "bT"]
cnt_cards = len(cards)
prob = 1/sc.comb(36, 2)
print(prob)
```

```python
#Проверим частотную вероятность
from random import randint

def DrawTwoCards(cnt):
    #Кол-во благоприятных исходов за cnt событий
    fav = 0
    x = []
    y = []
    for situation in range(1, cnt + 1):
        ind1 = randint(0, 35)
        ind2 = randint(0, 35)
        if (cards[ind1] == "bT") and (cards[ind2] == "bT") and (ind1 != ind2):
            fav += 1
        if situation % 1000 == 0:
            x.append(situation)
            y.append(fav/situation)

    fig, ax = plt.subplots()

    ax.plot(x, [1/sc.comb(36, 2)]*len(x), color = "blue")
    ax.plot(x, y, color = "red")

    
    print(f"Из {cnt} раз два чёрных туза попались {fav} раз. Доля выпавших двух чёрных тузов = {fav/cnt}")
```

```python
DrawTwoCards(100000)
```
