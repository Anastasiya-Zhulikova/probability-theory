> В конкурсе по 5 номинациям участвуют 10 кинофильмов. Сколько существует вариантов распределения призов, если по каждой номинации установлены: а) различные призы; б) одинаковые призы?

а) Т. к. призы различны, воспользуемся формулой размещения (с повторениями):
$\bar{A}_n^k = n^k = 10^5 = 100000$

```python
print(10**5)
```

б) Т. к. призы одинаковы, воспользуемся формулой сочетания (с повторениями):

```python
print(int(sc.comb(10+5-1, 5)))
```

### Задача 1.9
>Сколько существует семизначных чисел, состоящих из цифр 4, 5 и 6, в которых цифра 4 повторяется 3 раза, а цифры 5 и 6 - по 2 раза?

$* * * * * * *$
\
Воспользуемся формулой перемещения (с повторениями):
$P_n(n_1, n_2, ..., n_m) = \frac{n!}{n_1! \cdot n_2! \cdot ... \cdot n_m!} = \frac{7!}{3! \cdot 2! \cdot 2!} = 210$

```python
print(int(sc.factorial(7) / (sc.factorial(3) * sc.factorial(2) * sc.factorial(2))))
```

### Задача 1.17
>В купейный вагон (9 купе по 4 места) семи пассажирам продано 7 билетов. Найти вероятности того, что пассажиры попали: а) в два купе; б) в семь купе; в) в три купе.

Будем рассматривать отдельно выбор купе, отдельно выбор места в купе.
\
Все исходы: сочетания (без повторений) $C_n^k = \frac{n!}{k!(n-k)!} = \frac{36!}{7!(36-7)!} = 8347680$

```python
print(int(sc.comb(9, 2)))
```

```python
print(int(sc.comb(36, 7)))
```

\
Купе: повторы: нет. Порядок: нет. Значит воспользуемся формулой сочетания (без повторений): $C_n^k = \frac{n!}{k!(n-k)!} = \frac{9!}{2!(9-2)!} = 36$

```python
print(int(sc.comb(9, 2)))
```

С помощью той же формулы сочетания (без повторений) $C_n^k = \frac{n!}{k!(n-k)!} = \frac{8!}{7!(8-7)!}$ = 8 рассадим пассажиров:

```python
print(int(sc.comb(8, 7)))
```

Перемножим благоприятные исходы и поделим на все. Ответ: 

```python
(8 * 36)/(8347680)
```

б) Рассмотрим событие $B$ - семь пассажиров попали в семь купе (из 9 по 4 места).
\
Купе:

```python
print(int(sc.comb(9, 7)))
```

Одного человека в вагоне можно посадить на 4 места. Возпользуемся формулой размещения (с повторениями): $\bar{A}_n^k = n^k = 4^7 = 16384$

### Задача 1.48
>В старинной игре в кости необходимо было для выигрыша получить при бросании трех игральных костей сумму очков, превосходящую 10. Найти вероятности: а) выпадения ll очков; б) выигрыша.

a) Рассмотрим событие $A$ - на кости выпало 11 очков.
\
Все исходы: 6 * 6 * 6 = 216.
\
Переберём все благоприятные исходы и поспользуемся для них формулой перестановок (без повторений): $P_n = n!$
\
или перестановок (с повторениями): $P_n(n_1, n_2, ..., n_m) = \frac{n!}{n_1! \cdot n_2! \cdot ... \cdot n_m!}$
\
\
1 6 4 $P_n = 3! = 6$
\
2 5 4 $P_n = 3! = 6$
\
3 4 4 $P_n = 3!/2! = 3$
\
1 5 5 $P_n = 3!/2! = 3$
\
5 3 3 $P_n = 3!/2! = 3$
\
6 2 3 $P_n = 3! = 6$

Сложим благоприятные исходы и разделим на общее кол-во:

```python
print((6 + 6 + 3 + 3 + 3 + 6)/216)
```

Программный вариант решения:

```python
from itertools import product

#Кол-во благоприятных исходов
cnt = 0
ls = list(product("123456", repeat=3))
for ls_i in ls:
    if int(ls_i[0]) + int(ls_i[1]) + int(ls_i[2]) == 11:
        cnt += 1

#Вероятность
print(cnt/6**3)
```

Отсюда вытекает свойство: <b>число способов получить любую сумму S всегда равно числу способов получить симметричную ей сумму 21 - S</b>.

```python
from itertools import product

#Кол-во благоприятных исходов
cnt = 0
ls = list(product("123456", repeat=3))
for ls_i in ls:
    if int(ls_i[0]) + int(ls_i[1]) + int(ls_i[2]) >= 11:
        cnt += 1

#Вероятность
print(cnt/6**3)
```
